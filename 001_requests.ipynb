````python
# Create an API client

from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"

````

````python
def add_user_message(messages, text):
    user_message = {"role":'user',"content":text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant","content":text}
    messages.append(assistant_message)


def chat(messages,system=None,stop_sequences=[]):
    
    params = {
        "model" : model,
        "max_tokens" : 1000,
        "messages" : messages,
        "stop_sequences":stop_sequences,
    }

    if system:
        params["system"]=system

    message = client.messages.create(**params)

    return message.content[0].text


````

````python
# messages starting list
messages =[]
system_prompt = """
You are an expert programmer. you will write the code as consice as possible
"""

while(True):
    userInput = input()
    add_user_message(messages,userInput)
    #pass the list of messages to the claude from the chat funtion
    answer = chat(messages, system_prompt)
    #take the answer and add it to the assistant message into our list
    add_assistant_message(messages,answer)

    print(f"----------------------")
    print(f"< {userInput}")
    print(f"{answer}")
    print(f"----------------------")
````
